# overall apple analysis

```python
data = {
    "Product Name": [],
    "Review & Ratings": [],
    "Rating Number": [],
    "Memory": [],
    "Display": [],
    "Camera": [],
    "Processor": [],
    "Water Resistance": [],
    "Fast Charging": [],
    "Wireless Charging": [],
    "Warranty": [],
    "Actual Price": [],
    "Sale Price": [],
    "Offer": [],
}
```

```python
for i in range(2,27):
    url = "https://www.flipkart.com/search?q=mobile&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=off&as=off&p%5B%5D=facets.brand%255B%255D%3DApple&sort=price_asc&page=" + str(i)
    r = requests.get(url)
    soup = BeautifulSoup(r.text, "lxml")

    # list of product cards on THIS page
    all_info = soup.find_all("div", class_="tUxRFH")  # adjust if needed
    
    for card in all_info:          # I call it 'card' instead of 'contain'
    
        # -------- Product name --------
        name_tag = card.find("div", class_="KzDlHZ")
        product_name = name_tag.get_text(strip=True) if name_tag else None
        data["Product Name"].append(product_name)
    
        # -------- Rating & reviews --------
        # rr is a single tag, not a list → use find(), not find_all()
        rr = card.find("span", class_="Wphh3N")   # e.g. "4.5★ 1,54,623 Ratings & 11,955 Reviews"
        if rr is None:
            review_and_rating = None
            rating_number = None
        else:
            review_and_rating = rr.get_text(strip=True)
    
            # number of ratings only (if available)
            rating_num_tag = card.find("div", class_="XQDdHH")
            rating_number = rating_num_tag.get_text(strip=True) if rating_num_tag else None
    
        data["Review & Ratings"].append(review_and_rating)
        data["Rating Number"].append(rating_number)
    
        # -------- Specifications (bullets) --------
        spec_block = card.find("div", class_="_6NESgJ")
    
        # default: everything None
        memory = display = camera = processor = None
        water_res = fast_chg = wireless_chg = warranty = None
    
        if spec_block:
            li_texts = [li.get_text(strip=True) for li in spec_block.find_all("li")]
    
            # classify each bullet into the right column
            for text in li_texts:
                t = text.lower()
    
                if "rom" in t:
                    memory = text
                elif "display" in t:
                    display = text
                elif "camera" in t:
                    camera = text
                elif "bionic" in t or "chip" in t or "processor" in t:
                    processor = text
                elif "water" in t or "ip67" in t or "ip68" in t:
                    water_res = text
                elif "fast charge" in t or "fast charging" in t:
                    fast_chg = text
                elif "wireless charging" in t:
                    wireless_chg = text
                elif "warranty" in t:
                    warranty = text
    
        # add specs to data
        data["Memory"].append(memory)
        data["Display"].append(display)
        data["Camera"].append(camera)
        data["Processor"].append(processor)
        data["Water Resistance"].append(water_res)
        data["Fast Charging"].append(fast_chg)
        data["Wireless Charging"].append(wireless_chg)
        data["Warranty"].append(warranty)
    
        # -------- Actual price --------
        actual_price_tag = card.find("div", class_="yRaY8j ZYYwLA")
        my_price = actual_price_tag.get_text(strip=True) if actual_price_tag else None
        data["Actual Price"].append(my_price)
    
        # -------- Sale price --------
        sale_tag = card.find("div", class_="Nx9bqj _4b5DiR")
        sale_price = sale_tag.get_text(strip=True) if sale_tag else None
        data["Sale Price"].append(sale_price)
    
        # -------- Offer text --------
        offer_tag = card.find("div", class_="UkUFwK")
        if offer_tag:
            data["Offer"].append(offer_tag.get_text(strip=True))
        else:
            data["Offer"].append("no offer")
```

```python
for i in range(1,44):
    my_url_samsung = "https://www.flipkart.com//search?q=mobile&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&sort=price_asc&p%5B%5D=facets.brand%255B%255D%3DSamsung&page=" + str(i)
    my_r_samsung = requests.get(my_url_samsung)
    my_soup_samsung = BeautifulSoup(my_r_samsung.text, "lxml")

    # list of product cards on THIS page
    all_info_samsung = my_soup_samsung.find_all("div", class_="yKfJKb row")  # adjust if needed
    
    for card_samsung in all_info_samsung:          # I call it 'card' instead of 'contain'
    
        # -------- Product name --------
        name_tag_samsung = card_samsung.find("div", class_="KzDlHZ")
        product_name_samsung = name_tag_samsung.get_text(strip=True) if name_tag_samsung else None
        data_samsung["Product Name"].append(product_name_samsung)

    
        # -------- Rating & reviews --------
        # rr is a single tag, not a list → use find(), not find_all()
        rr_samsung = card_samsung.find("span", class_="Wphh3N")   # e.g. "4.5★ 1,54,623 Ratings & 11,955 Reviews"
        if rr_samsung is None:
            review_and_rating_samsung = None
            rating_number_samsung= None
        else:
            review_and_rating_samsung = rr_samsung.get_text(strip=True)
    
            # number of ratings only (if available)
            rating_num_tag_samsung = card_samsung.find("div", class_="XQDdHH")
            rating_number_samsung = rating_num_tag_samsung.get_text(strip=True) if rating_num_tag_samsung else None
    
        data_samsung["Review & Ratings"].append(review_and_rating_samsung)
        data_samsung["Rating Number"].append(rating_number_samsung)
    
        # -------- Specifications (bullets) --------
        spec_block_samsung = card_samsung.find("div", class_="_6NESgJ")
    
        # default: everything None
        memory_samsung = display_samsung = camera_samsung = processor_samsung = None
        water_res_samsung = fast_chg_samsung = wireless_chg_samsung = warranty_samsung = battery_samsung =None
    
        if spec_block_samsung:
            li_texts_samsung = [li.get_text(strip=True) for li in spec_block_samsung.find_all("li")]
    
            # classify each bullet into the right column
            for text in li_texts_samsung:
                t_samsung = text.lower()
    
                if "rom" in t_samsung or "ram" in t_samsung:
                    memory_samsung = text
                elif "display" in t_samsung:
                    display_samsung = text
                elif "camera" in t_samsung:
                    camera_samsung = text
                elif "bionic" in t_samsung or "chip" in t_samsung or "processor" in t_samsung:
                    processor_samsung = text
                elif "water" in t_samsung or "ip67" in t_samsung or "ip68" in t_samsung:
                    water_res_samsung = text
                elif "fast charge" in t_samsung or "fast charging" in t_samsung:
                    fast_chg_samsung = text
                elif "wireless charging" in t_samsung:
                    wireless_chg_samsung = text
                elif "warranty" in t_samsung or "year" in t_samsung or "months" in t_samsung:
                    warranty_samsung = text
                elif "battery" in t_samsung or "mah" in t_samsung:
                    battery_samsung = text
        data_samsung["Memory"].append(memory_samsung)
        data_samsung["Display"].append(display_samsung)
        data_samsung["Camera"].append(camera_samsung)
        data_samsung["Processor"].append(processor_samsung)
        data_samsung["Water Resistance"].append(water_res_samsung)
        data_samsung["Fast Charging"].append(fast_chg_samsung)
        data_samsung["Wireless Charging"].append(wireless_chg_samsung)
        data_samsung["Warranty"].append(warranty_samsung)
        data_samsung["Battery"].append(battery_samsung)
           
        
        
    
        # -------- Actual price --------
        actual_price_tag_samsung = card_samsung.find("div", class_="yRaY8j ZYYwLA")
        my_price_samsung = actual_price_tag_samsung.get_text(strip=True) if actual_price_tag_samsung else None
        data_samsung["Actual Price"].append(my_price_samsung)
    
        # -------- Sale price --------
        sale_tag_samsung = card_samsung.find("div", class_="Nx9bqj _4b5DiR")
        sale_price_samsung = sale_tag_samsung.get_text(strip=True) if sale_tag_samsung else None
        data_samsung["Sale Price"].append(sale_price_samsung)
        # -------- Offer text --------
        offer_tag_samsung = card_samsung.find("div", class_="UkUFwK")
        offer_samsung = offer_tag_samsung.get_text(strip=True) if offer_tag_samsung else "no offer"
        data_samsung["Offer"].append(offer_samsung)
       
        
  
```
